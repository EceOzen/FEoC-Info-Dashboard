# CMIP6 vs CMIP7 volcanic forcing (synthetic example)

```python
import math
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
years = list(range(1950, 2024))
def pulse(y, shift=0.0):
    e = {1963: 0.7, 1982: 0.6, 1991: 1.1}
    return -sum(a*3*math.exp(-(y-t)/(1.2+shift)) for t, a in e.items() if y >= t)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(years, [pulse(y) for y in years], label='CMIP6 (synthetic)', color='#5C7A85')
ax.plot(years, [pulse(y, 0.3) for y in years], label='CMIP7 (synthetic)', color='#B5563C')
ax.set_ylabel('ERF (W m$^{-2}$)'); ax.legend(frameon=False)
ax.spines[['top','right']].set_visible(False)
plt.show()
```
